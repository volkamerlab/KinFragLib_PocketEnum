```python
from rdkit.Chem import PandasTools 
import pandas as pd
import matplotlib.pyplot as plt
from rdkit import Chem

from src.evaluation.utils import read_mols, standardize_mol
```

```python
enumerated = PandasTools.LoadSDF("../results_compare_full/5n1f_enumerated/full_enumeration_docked_scored_ap_fp_dedupl.sdf")
enumerated.head()
```

```python
data = read_mols('../results_5n1f_25_02/5n1f/results.sdf')
```

```python
data.head()
```

```python
_enumerated = PandasTools.LoadSDF("../results_compare_full/5n1f_enumerated/full_enumeration_docked_scored_dedupl.sdf")
_enumerated.head()
```

```python
enumerated = pd.concat([enumerated, _enumerated], ignore_index=True)
```

```python
enumerated["estimated_affinity"] = enumerated.apply(lambda r: (float(r["BIOSOLVEIT.HYDE_ESTIMATED_AFFINITY_LOWER_BOUNDARY [nM]"]) + 
                                                               float(r["BIOSOLVEIT.HYDE_ESTIMATED_AFFINITY_UPPER_BOUNDARY [nM]"]))/2, axis=1)
```

```python
fragment_based = PandasTools.LoadSDF("../results_compare_full/5n1f_pipeline/results.sdf")
fragment_based.head()
```

```python
fragment_based["estimated_affinity"] = fragment_based.apply(lambda r: (float(r["BIOSOLVEIT.HYDE_ESTIMATED_AFFINITY_LOWER_BOUNDARY [nM]"]) + 
                                                               float(r["BIOSOLVEIT.HYDE_ESTIMATED_AFFINITY_UPPER_BOUNDARY [nM]"]))/2, axis=1)
```

```python
fragment_based = fragment_based.sort_values("estimated_affinity", ignore_index=True)
fragment_based = fragment_based.drop_duplicates(subset=["inchi"])
len(fragment_based)
```

```python
enumerated = enumerated.sort_values("estimated_affinity", ignore_index=True)
enumerated = enumerated.drop_duplicates(subset=["inchi"])
len(enumerated)
```

```python
plt.hist((enumerated[enumerated["estimated_affinity"] < 100000])["estimated_affinity"],  histtype='step', label="Fully enumerated")
plt.hist((fragment_based[fragment_based["estimated_affinity"] < 100000])["estimated_affinity"],  histtype="step", label="Fragment-based")
plt.legend()
plt.ylabel("#Compounds")
plt.show()
```

```python
print("Top 25 ranked compounds")
mols = [enumerated.ROMol[idx] for idx in range(10)]
for mol in mols:
    Chem.AllChem.Compute2DCoords(mol)
img = Chem.Draw.MolsToGridImage(
    mols, 
    legends=[
        f"{round(enumerated['estimated_affinity'][idx],4)} nM"  
        for idx in range(10)
    ],  
    molsPerRow=5, 
    maxMols=100,
    subImgSize=(200, 200)
)
img
```
